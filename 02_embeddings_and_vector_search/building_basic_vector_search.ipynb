# 🧠 Building a Basic Vector Search System

This notebook walks through **how vector search actually works**, step by step.

You will learn:
- How text becomes embeddings
- How vectors are stored
- How similarity search retrieves candidates
- Where vector search succeeds
- Where it fundamentally fails

This is not a framework demo.
This is **mechanics and intuition**.


## 1. What We Are Building

We will build a minimal vector search system:

1. Take documents
2. Convert them to embeddings
3. Store embeddings
4. Retrieve the most similar chunks for a query

No generation yet.
Only retrieval.


## 2. Why Start with Vector Search Alone

Vector search is:
- the foundation of RAG
- where most failures originate
- independent of LLM generation

If retrieval is wrong,
generation cannot fix it.


## 3. Sample Documents

We will use a small corpus to illustrate behavior.

Each document should represent:
- one coherent idea
- minimal noise


```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = [
    "Paris is the capital of France.",
    "Berlin is the capital of Germany.",
    "The Eiffel Tower is located in Paris.",
    "Machine learning enables computers to learn from data.",
    "Deep learning is a subset of machine learning."
]
```

## 6. What an Embedding Function Represents

An embedding function maps text to vectors such that:
- similar meanings → similar vectors
- unrelated meanings → distant vectors

We will simulate embeddings to focus on retrieval logic.


```python
np.random.seed(42)

def fake_embedding(text, dim=8):
    return np.random.rand(dim)

document_embeddings = np.array([fake_embedding(doc) for doc in documents])
```

```python
query = "What is the capital of France?"
query_embedding = fake_embedding(query)
```

```python
similarities = cosine_similarity(
    query_embedding.reshape(1, -1),
    document_embeddings
)[0]
```

```python
results = list(zip(documents, similarities))
results = sorted(results, key=lambda x: x[1], reverse=True)

for doc, score in results:
    print(f"{score:.3f} | {doc}")
```

## 11. What Just Happened?

We performed:

- embedding(query)
- embedding(documents)
- similarity(query, document)
- ranking by similarity

This is **vector search**.


## 12. What Vector Search Is Good At

- Semantic similarity
- Topic matching
- Paraphrase detection
- Fuzzy recall

It answers:
> “What is related?”


## 13. What Vector Search Is Bad At

- Exact facts
- Numbers
- Time-based correctness
- Permissions
- Logic

It cannot answer:
> “What is correct?”


## 14. Why This Alone Is Not RAG

RAG requires:
- correct retrieval
- grounding
- controlled context injection
- generation with constraints

Vector search is only:
> candidate selection


## 15. A Typical Failure Mode

If the corpus contains:
- outdated docs
- loosely related content
- mixed domains

Vector search will still retrieve them.

Similarity ≠ validity.


## Final Mental Lock

Vector search answers:
> “What looks similar?”

It does NOT answer:
> “What is true?”

Never confuse retrieval with reasoning.
