# 🧠 JSON Schema & Function Calling

This notebook explains **how JSON schemas and function calling
transform LLMs from text generators into reliable system components**.

You will learn:
- Why free-form text breaks systems
- What JSON schema actually guarantees (and what it doesn’t)
- How function calling really works
- Common failure modes in schema-based systems
- How to design schemas that survive production

📌 Core idea:
> Language is probabilistic.
> Schemas are contracts.


## 1. The Core Problem

LLMs naturally produce:
- fluent text
- variable formats
- creative structures

Systems require:
- strict structure
- predictable fields
- machine-readable output

Free-form text cannot be trusted by code.


## 2. What Is JSON Schema?

JSON Schema is:
> A formal contract that defines
> the structure, types, and constraints of JSON data

It specifies:
- required fields
- allowed values
- data types
- nesting rules


## 3. What JSON Schema Is NOT

JSON Schema does NOT:
- guarantee correctness
- guarantee truth
- guarantee business logic validity

It guarantees:
> structural correctness only


## 4. Schema-First Design

In production systems:
- output shape matters more than prose quality
- missing fields cause crashes
- malformed data causes silent bugs

Schema-first design shifts failure:
- from silent → explicit


## 5. What Is Function Calling?

Function calling allows an LLM to:
- choose a tool/function
- return arguments in structured JSON
- defer execution to deterministic code

The model decides:
> WHAT function to call and with WHAT inputs

The system decides:
> HOW the function executes


## 6. Function Calling vs “Please Output JSON”

Prompted JSON:
- brittle
- breaks under pressure
- fails silently

Function calling:
- schema-validated
- constrained
- machine-safe

Never rely on “please output JSON” in production.


## 7. Minimal Example Schema

Example task:
"Extract user intent from a query"


```python
intent_schema = {
    "name": "classify_intent",
    "description": "Classify user intent",
    "parameters": {
        "type": "object",
        "properties": {
            "intent": {
                "type": "string",
                "enum": ["search", "summarize", "compare", "unknown"]
            },
            "confidence": {
                "type": "number",
                "minimum": 0,
                "maximum": 1
            }
        },
        "required": ["intent", "confidence"]
    }
}
```

## 8. Model Output (Conceptual)

Instead of text, the model outputs:

{
  "intent": "search",
  "confidence": 0.87
}

This output is:
- parseable
- enforceable
- debuggable


## 9. Hallucination Reduction

Schemas reduce hallucination by:
- eliminating creative freedom in structure
- forcing explicit uncertainty
- preventing invented fields

The model can still be wrong —
but it cannot be vague.


## 10. Schema Mistakes

❌ Overly flexible schemas  
❌ Optional fields everywhere  
❌ Free-text where enums should exist  
❌ Mixing business logic into schema  

Schemas should constrain, not describe.


## 11. Validation Is Mandatory

Schema without validation is useless.

Always:
- validate model output
- reject invalid payloads
- retry or fallback safely


## 12. Failure Modes

❌ Model selects wrong function  
❌ Correct schema, wrong values  
❌ Overconfidence in confidence scores  
❌ Schema drift over time  

Schemas reduce failure surface —
they do not eliminate it.


## 13. Schemas as Boundaries

Schemas define:
- what the model is allowed to say
- what the system is willing to accept

Anything outside schema:
> is rejected safely


## 14. Use Function Calling When:

- output feeds another system
- correctness matters
- actions have side effects
- decisions must be logged
- retries are needed


## 15. Avoid Function Calling When:

- output is purely conversational
- format flexibility is acceptable
- failure cost is low

Not everything needs structure.


## Final Mental Lock

Prompts influence probabilities.
Schemas enforce contracts.
Functions execute reality.

Production GenAI requires all three.
